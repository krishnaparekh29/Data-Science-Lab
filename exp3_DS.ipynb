**GROUPED ATTRIBUTES**

```python
# Grouped Data Analysis using two attributes:
# soil_type and seed_variety
# Statistical measures: Frequency, Mean, Median and Mode

#  Mean
def calculate_mean(data):
    total = 0

    for value in data:
        total = total + value

    return total / len(data)


#  Median
def calculate_median(data):
    sorted_data = sorted(data)
    n = len(sorted_data)

    if n % 2 == 0:
        middle1 = sorted_data[n // 2 - 1]
        middle2 = sorted_data[n // 2]
        return (middle1 + middle2) / 2
    else:
        return sorted_data[n // 2]


#  Mode
def calculate_mode(data):
    frequency = {}

    for value in data:
        if value in frequency:
            frequency[value] = frequency[value] + 1
        else:
            frequency[value] = 1

    highest_frequency = 0

    for value in frequency:
        if frequency[value] > highest_frequency:
            highest_frequency = frequency[value]

    # No repeated value
    if highest_frequency == 1:
        return "No mode"

    modes = []

    for value in frequency:
        if frequency[value] == highest_frequency:
            modes.append(value)

    return modes


# Group dataset by soil type and seed variety
grouped_data = df.groupby(["soil_type", "seed_variety"])

# Calculate frequency, mean, median and mode
for (soil, seed), group in grouped_data:

    yield_data = group["yield_quintal_per_acre"].tolist()

    print("\n" + "=" * 60)
    print("Soil Type    :", soil)
    print("Seed Variety :", seed)
    print("Frequency    :", len(yield_data))
    print("Mean Yield   :", calculate_mean(yield_data))
    print("Median Yield :", calculate_median(yield_data))
    print("Mode Yield   :", calculate_mode(yield_data))
```

The dataset was grouped using two categorical attributes, soil type and seed variety, to analyze sugarcane yield under different cultivation combinations. Frequency, mean, median, and mode were calculated manually for each group. This helps compare productivity across different soil and seed-variety conditions and identify combinations associated with higher or more typical sugarcane yield.

CONVERTING A DISCRETE ATTRIBUTE INTO CONTINOUS

```python
import numpy as np


# Step 1: Convert discrete attribute to continuous


np.random.seed(42)

continuous_irrigation = (
    df["irrigation_frequency_per_month"].astype(float)
    + np.random.uniform(-0.49, 0.49, size=len(df))
)

df["continuous_irrigation_frequency"] = continuous_irrigation


# Step 2: Mean


def calculate_mean(data):
    total = 0

    for value in data:
        total = total + value

    return total / len(data)


# Step 3: Median

def calculate_median(data):
    sorted_data = sorted(data)
    n = len(sorted_data)

    if n % 2 == 0:
        middle1 = sorted_data[n // 2 - 1]
        middle2 = sorted_data[n // 2]

        return (middle1 + middle2) / 2

    else:
        return sorted_data[n // 2]


# Step 4: Mode


def calculate_mode(data):
    frequency = {}

    for value in data:

        if value in frequency:
            frequency[value] += 1

        else:
            frequency[value] = 1

    highest_frequency = 0

    for value in frequency:

        if frequency[value] > highest_frequency:
            highest_frequency = frequency[value]

    if highest_frequency == 1:
        return "No mode"

    modes = []

    for value in frequency:

        if frequency[value] == highest_frequency:
            modes.append(value)

    return modes


# Step 5: Apply calculations


continuous_data = df[
    "continuous_irrigation_frequency"
].tolist()

print("Continuous Irrigation Frequency")
print("=" * 45)

print("Mean   :", calculate_mean(continuous_data))
print("Median :", calculate_median(continuous_data))
print("Mode   :", calculate_mode(continuous_data))
```

The discrete irrigation_frequency_per_month attribute was converted into a continuous representation and analyzed .

**STANDARD DEVIATION AND VARIANCE**

```python
# Manual calculation
#  Variance
def calculate_variance(data):
    mean = calculate_mean(data)

    total = 0

    for value in data:
        total = total + (value - mean) ** 2

    # Sample variance
    variance = total / (len(data) - 1)

    return variance


#  Standard Deviation
def calculate_standard_deviation(data):
    variance = calculate_variance(data)

    return variance ** 0.5


# Apply to the selected attributes
attributes = [
    "rainfall_mm",
    "avg_temperature_celsius",
    "yield_quintal_per_acre"
]

for column in attributes:

    data = df[column].tolist()

    print("\nAttribute:", column)
    print("-" * 50)

    print(" Variance:",
          calculate_variance(data))

    print(" Standard Deviation:",
          calculate_standard_deviation(data))
```

```python
print("Variance and Standard Deviation using built in functions")
print("=" * 50)

for column in attributes:

    variance = df[column].var()
    standard_deviation = df[column].std()

    print("\nAttribute:", column)
    print("Variance:", variance)
    print("Standard Deviation:", standard_deviation)
```

we calculated variance and standard deviation using both manual and built in functions. It helps us determine the consistency of the agricultural conditions from their average values . High rainfall SD shows rainfall conditions vary considerably across fields.

High yield SD shows sugarcane productivity varies considerably.

**INTERQUARTILE RANGE**

```python
# IQR calculation

def calculate_median(data):
    data = sorted(data)
    n = len(data)

    if n % 2 == 0:
        return (data[n // 2 - 1] + data[n // 2]) / 2
    else:
        return data[n // 2]


def calculate_iqr(data):
    data = sorted(data)
    n = len(data)

    # Split data into lower and upper halves
    if n % 2 == 0:
        lower_half = data[:n // 2]
        upper_half = data[n // 2:]
    else:
        lower_half = data[:n // 2]
        upper_half = data[n // 2 + 1:]

    q1 = calculate_median(lower_half)
    q3 = calculate_median(upper_half)

    return q3 - q1


for column in attributes:

    data = df[column].tolist()

    print("\nAttribute:", column)
    print("Q1:", calculate_median(sorted(data)[:len(data)//2]))
    print("Q3:", calculate_median(sorted(data)[len(data)//2:]))
    print("IQR:", calculate_iqr(data))
```

```python
print("IQR using built-in functions")
print("=" * 50)

for column in attributes:

    q1 = df[column].quantile(0.25)
    q3 = df[column].quantile(0.75)

    iqr = q3 - q1

    print("\nAttribute:", column)
    print("Q1:", q1)
    print("Q3:", q3)
    print("IQR:", iqr)
```

The Interquartile Range is used to measure the spread of the middle 50% of observations. For rainfall, the middle 50% of values lie between 1005.03 mm and 1684.815 mm, giving an IQR of 679.785 mm, indicating considerable variation in typical rainfall conditions. For average temperature, the middle 50% lies between 28.49°C and 33.1425°C, with an IQR of 4.6525°C, showing relatively smaller variation in temperature. For sugarcane yield, the middle 50% lies between 306.145 and 503.81 quintal/acre, giving an IQR of 197.665 quintal/acre, indicating noticeable variation in typical crop productivity.

IQR is useful for the irrigation advisory system because it focuses on the central 50% of field conditions and is less affected by extreme values. The results show that rainfall has the greatest spread among the selected attributes, while temperature is comparatively stable. The substantial spread in yield indicates that sugarcane productivity varies considerably even within the central portion of the dataset, supporting the need for irrigation recommendations that account for differing field conditions.

```python
attributes = [
    "rainfall_mm",
    "avg_temperature_celsius",
    "yield_quintal_per_acre"
]

for column in attributes:

    q1 = df[column].quantile(0.25)
    q3 = df[column].quantile(0.75)

    iqr = q3 - q1

    # Mild outlier limits
    mild_lower = q1 - 1.5 * iqr
    mild_upper = q3 + 1.5 * iqr

    # Extreme outlier limits
    extreme_lower = q1 - 3 * iqr
    extreme_upper = q3 + 3 * iqr

    # Counts
    mild_outliers = (
        (df[column] < mild_lower) |
        (df[column] > mild_upper)
    ).sum()

    extreme_outliers = (
        (df[column] < extreme_lower) |
        (df[column] > extreme_upper)
    ).sum()

    print("\n" + "=" * 60)
    print("Attribute:", column)
    print("Q1:", q1)
    print("Q3:", q3)
    print("IQR:", iqr)

    print("\nMild Outlier Limits:")
    print("Lower:", mild_lower)
    print("Upper:", mild_upper)
    print("Number of Mild Outliers:", mild_outliers)

    print("\nExtreme Outlier Limits:")
    print("Lower:", extreme_lower)
    print("Upper:", extreme_upper)
    print("Number of Extreme Outliers:", extreme_outliers)
```

The IQR method was used to detect mild and extreme outliers in rainfall, average temperature, and sugarcane yield. The analysis found no mild or extreme outliers in any of the three attributes. All 5,000 observations fall within the calculated outlier limits.

**GROUPED ANALYSIS OF VARIANCE, SD AND IQR**

```python
grouped_data = df.groupby(["soil_type", "seed_variety"])

for (soil, seed), group in grouped_data:

    yield_data = group["yield_quintal_per_acre"]

    variance = yield_data.var()
    standard_deviation = yield_data.std()

    q1 = yield_data.quantile(0.25)
    q3 = yield_data.quantile(0.75)
    iqr = q3 - q1

    print("\n" + "=" * 60)
    print("Soil Type    :", soil)
    print("Seed Variety :", seed)
    print("Frequency    :", len(yield_data))
    print("Variance     :", variance)
    print("Standard Deviation:", standard_deviation)
    print("IQR          :", iqr)
```

The grouped analysis shows that sugarcane yield variability differs substantially depending on the combination of soil type and seed variety. Some combinations, such as Alluvial–Co86032 and Clay Loam–Co0238, show large variation in productivity, whereas combinations such as Red Soil–CoM0265 and Sandy Loam–CoM0265 show much more consistent yields. This demonstrates that agricultural conditions are not uniform across fields. For an AI-based irrigation advisory system, this supports the need for condition-specific or plot-specific recommendations rather than applying one irrigation strategy to every sugarcane field
