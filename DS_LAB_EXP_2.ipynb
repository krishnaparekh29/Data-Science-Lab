Correlation analysis was performed to determine the relationship between rainfall and sugarcane yield. This shows that in this dataset, higher rainfall is generally associated with higher sugarcane yield. This relationship is useful for the irrigation advisory use case because rainfall is an important factor when determining irrigation requirements.

```python
normalized_df = df.copy()

cols = [
    "rainfall_mm",
    "avg_temperature_celsius",
    "soil_moisture_percent"
]

normalized_df[cols] = (
    normalized_df[cols] - normalized_df[cols].min()
) / (
    normalized_df[cols].max() - normalized_df[cols].min()
)

normalized_df[cols].head()
```

Thus normalization was performed on rainfall, average temperature, and soil moisture to bring the three attributes to a common scale between 0 and 1. The output shows that the original values have been transformed into normalized values . This transformation makes the numerical attributes comparable despite having different units and ranges.

```python
df_left = df[
    ["soil_type", "rainfall_mm", "yield_quintal_per_acre"]
].head(5)

df_right = df[
    ["water_requirement_liters_per_day",
     "irrigation_frequency_per_month"]
].head(5)

joined_df = df_left.join(df_right)

joined_df
```

Joining was performed to combine environmental, crop-yield, and irrigation related attributes from two DataFrames using their common row index. The output combines soil type, rainfall, yield, water requirement, and irrigation frequency .

```python
merge_left = df[
    ["soil_type", "rainfall_mm", "yield_quintal_per_acre"]
].head(10).copy()

merge_right = df[
    ["water_requirement_liters_per_day",
     "irrigation_frequency_per_month"]
].head(10).copy()

merge_left["record_id"] = range(1, 11)
merge_right["record_id"] = range(1, 11)

merged_df = pd.merge(
    merge_left,
    merge_right,
    on="record_id"
)

merged_df
```

Merging was performed to combine agricultural and irrigation related information using record_id as a common key.
